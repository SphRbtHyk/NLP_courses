# 2. Lire un fichier texte

Un fichier texte contient des caractères encodés sous forme d'octets.

Nous utiliserons toujours :

- `with open(...)` pour garantir la fermeture du fichier ;
- `encoding="utf-8"` pour lire correctement les accents.

```python
with open("donnees/poeme.txt", mode="r", encoding="utf-8") as fichier:
    contenu = fichier.read()

print(type(contenu))
print(contenu[:150])
```

## Trois manières courantes de lire

| Instruction | Résultat | Usage |
|---|---|---|
| `fichier.read()` | une chaîne | lire tout le fichier |
| `fichier.readlines()` | une liste de lignes | conserver les retours à la ligne |
| `for ligne in fichier:` | une ligne à la fois | traiter un fichier volumineux |

```python
with open("donnees/poeme.txt", encoding="utf-8") as fichier:
    lignes = fichier.readlines()

print(type(lignes))
print("Nombre d'éléments :", len(lignes))
print("Première ligne :", repr(lignes[0]))
```

# 3. Écrire un fichier texte

Le mode choisi détermine ce que devient le fichier.

| Mode | Effet |
|---|---|
| `"w"` | crée ou **remplace** le fichier |
| `"a"` | ajoute à la fin du fichier |
| `"r"` | lit le fichier ; mode par défaut |

<div style="background:#ffeaea; border-left:6px solid #b22222; padding:14px 18px; margin:12px 0; border-radius:4px;">
<strong>⚠️</strong><br>Le <code>mode='w'</code> <b>écrase</b> le contenu précédent (perte de données).
</div>

```python
resume = f"Le poème contient {len(lignes)} lignes et {len(mots)} mots.\n"

with open("resultats/resume_poeme.txt", mode="w", encoding="utf-8") as fichier:
    fichier.write(resume)

with open("resultats/resume_poeme.txt", encoding="utf-8") as fichier:
    print(fichier.read())
```

# 4. Le format CSV

Un fichier **CSV** représente un tableau en texte brut :

- une ligne par observation ;
- une colonne par variable ;
- un séparateur entre les colonnes.

En France, le séparateur est très souvent le **point-virgule** (virgule=séparateur décimal en français).

```python
with open("donnees/enquete.csv", encoding="utf-8") as fichier:
    for _ in range(5):
        print(fichier.readline().strip())
```

On utilise pour lire un csv le modèle `csv`, qui permet de gérer les séparateurs, les guillemets et les retours à la ligne.

```python
import csv

with open("donnees/enquete.csv", encoding="utf-8", newline="") as fichier:
    lecteur = csv.DictReader(fichier, delimiter=";")
    premiere_ligne = next(lecteur)

print(premiere_ligne)
print(premiere_ligne["niveau"])
print(type(premiere_ligne["age"]))
```

<div style="background:#e8f7f6; border-left:6px solid #159e9a; padding:14px 18px; margin:12px 0; border-radius:4px;">
La lib `csv` fait que toutes les valeurs sont lues comme des <code>str</code>. Pour calculer, il faut convertir explicitement avec <code>int()</code> ou <code>float()</code>.
</div>

## Compter par catégorie

On peut utiliser un dictionnaire pour associer à chaque modalité à son effectif.

```python
effectifs = {}

with open("donnees/enquete.csv", encoding="utf-8", newline="") as fichier:
    lecteur = csv.DictReader(fichier, delimiter=";")
    for ligne in lecteur:
        niveau = ligne["niveau"]
        if niveau not in effectifs:
            effectifs[niveau] = 0
        effectifs[niveau] += 1

print(effectifs)
```

# 6. Quel format choisir ?

| Besoin | Format adapté |
|---|---|
| texte continu, corpus, notes | TXT |
| tableau d'observations homogènes | CSV |
| données imbriquées ou résultat structuré | JSON |

Le format dépend de la **structure des données** et de l'usage prévu.
