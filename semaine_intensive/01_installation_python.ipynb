# Installer et prendre en main son environnement

**Objectifs** :

- retrouver le dossier du cours depuis un terminal ;
- créer et activer un environnement virtuel ;
- installer et lancer JupyterLab ;
- créer, exécuter et enregistrer un notebook.

⚠️ On ne continue pas si on est bloqué!

## Les quatre éléments à ne pas confondre

| Élément | Rôle |
|---|---|
| **Terminal** | La fenêtre dans laquelle nous écrivons des commandes |
| **Shell** | Le programme qui lit et exécute ces commandes |
| **Python** | Le programme qui exécute le code Python |
| **JupyterLab** | L'interface dans laquelle nous créons des notebooks |

JupyterLab ne remplace pas Python : il transmet à Python le contenu des cellules de code.

## Avant de commencer : repérez votre système

Les commandes ne sont pas toujours identiques. Suivez seulement la colonne correspondant à votre ordinateur.

| Windows | macOS ou Linux |
|---|---|
| Ouvrir **Invite de commandes** ou **Powershell** (`cmd`) | Ouvrir **Terminal** |
| Commande Python : `py` avant activation | Commande Python : `python3` avant activation |
| Séparateur de dossier : `\` | Séparateur de dossier : `/` |


## Créer le dossier du cours

## Le shell : savoir où l'on se trouve

Dans le terminal, entrez une commande puis appuyez sur **Entrée**.

| Action | Windows | macOS / Linux |
|---|---|---|
| afficher le dossier actuel | `cd` | `pwd` |
| afficher son contenu | `dir` | `ls` |
| entrer dans un dossier | `cd nom_du_dossier` | `cd nom_du_dossier` |
| remonter d'un dossier | `cd ..` | `cd ..` |
| créer un dossier | `mkdir nom` | `mkdir nom` |

>
> On peut compléter la commande grâce à la touche **Tabulation**. C'est plus rapide et cela évite les fautes de frappe.

## Comprendre un chemin

Un chemin indique où se trouve un fichier ou un dossier.

```text
Documents / python_SHS / seance_01 / essai.ipynb
```

- un **chemin absolu** part de la racine du disque ;
- un **chemin relatif** part du dossier où l'on se trouve ;
- `.` désigne le dossier actuel ;
- `..` désigne le dossier parent.

La plupart des erreurs « fichier introuvable » sont en réalité des erreurs de chemin.

## Vérifier que Python est installé

Placez-vous dans `python_SHS`, puis entrez **une seule** des commandes suivantes.

**Windows**


**macOS ou Linux**

> Ne poursuivez pas tant que la commande de vérification ne renvoie pas un numéro de version!!!

## Pourquoi créer un environnement virtuel ?

Un environnement virtuel est une installation Python isolée dans le dossier du projet.

```text
python_SHS/
├── .venv/        ← Python et bibliothèques du cours
├── seance_01/
└── ...
```

Ainsi, les bibliothèques installées pour ce cours ne perturbent ni Python du système ni les autres projets.

> Le dossier `.venv` est généré automatiquement. Ne modifiez jamais son contenu à la main.

## Créer l'environnement `.venv`

Vérifiez d'abord que le terminal se trouve dans `python_SHS`. Puis :

**Windows**

```bat
py -m venv .venv
```

**macOS ou Linux**

```bash
python3 -m venv .venv
```

La commande peut ne rien afficher : l'absence de message d'erreur est normale. Avec `dir` ou `ls -a`, vérifiez ensuite que `.venv` existe.

## Activer l'environnement

**Windows — Invite de commandes**

```bat
.venv\Scripts\activate.bat
```

**macOS ou Linux**

```bash
source .venv/bin/activate
```

Le début de la ligne doit maintenant contenir `(.venv)`. À partir de ce moment, la commande `python` désigne Python de l'environnement virtuel.

## Vérifier l'activation

Dans **tous les systèmes**, exécutez maintenant :

```bash
python --version
python -c "import sys; print(sys.executable)"
```

La seconde commande doit afficher un chemin qui contient à la fois :

- le dossier `python_SHS` ;
- le dossier `.venv`.

Si ce n'est pas le cas, arrêtez-vous : l'environnement n'est pas correctement activé.

## Installer JupyterLab

L'environnement doit toujours être actif : vérifiez la présence de `(.venv)`.

```bash
python -m pip install --upgrade pip
python -m pip install jupyterlab
```

- `pip` installe des bibliothèques Python ;
- `python -m pip` garantit que l'installation se fait dans le Python actif ;
- de nombreuses lignes peuvent s'afficher ;
- attendez le retour de l'invite avant d'entrer une nouvelle commande.

## Lancer JupyterLab

Toujours depuis `python_SHS` et avec `(.venv)` visible :

```bash
python -m jupyter lab
```

Une page doit s'ouvrir dans le navigateur. Le terminal reste occupé : c'est normal, JupyterLab fonctionne tant que ce processus reste actif.

> Ne fermez pas le terminal pendant que vous travaillez dans JupyterLab.

- une cellule **Code** contient des instructions exécutées par Python ;
- une cellule **Markdown** contient du texte, des titres, des liens ou des consignes.

Raccourcis importants :

| Action | Raccourci |
|---|---|

| enregistrer | `Ctrl+S` / `Cmd+S` |
| interrompre un calcul | bouton carré **Stop** |

## Ordre d'exécution

> ⚠️ Dans un notebook, vous pouvez exécuter les cellules dans n'importe quel ordre **ce qui peut créer des confusions**.

1. Exécutez une cellule contenant `nombre = 10`.
2. Plus bas, exécutez `nombre + 5`.
3. Modifiez la première cellule en `nombre = 100`, **sans l'exécuter**.
4. Réexécutez seulement `nombre + 5`.

Pourquoi le résultat vaut-il encore `15` ?

## Le noyau conserve l'état du calcul

Le **kernel** ou **noyau** est le processus Python qui exécute les cellules et garde les variables en mémoire.

Pour vérifier qu'un notebook fonctionne vraiment :

1. ouvrez le menu **Kernel** ;
2. choisissez **Restart Kernel and Run All Cells** ;
3. vérifiez qu'aucune cellule ne produit d'erreur.

> À toujours faire avant de rendre un notebook.

## Lire un message d'erreur

Exécutez volontairement :

````python
print(bonjour)
````

Un message d'erreur est **une information à lire de bas en haut**:

- dernière ligne : le type d'erreur et son explication ;
- ligne signalée : l'instruction qui a provoqué l'erreur ;
- ici, `bonjour` est interprété comme un nom de variable qui n'a jamais été défini.

Pour afficher du texte, il faut écrire `print("bonjour")`.
