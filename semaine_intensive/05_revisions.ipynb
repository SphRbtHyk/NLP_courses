## Compter les occurrences

Pour chaque valeur :

1. si elle est déjà dans le dictionnaire, augmenter son compteur ;
2. sinon, créer son compteur avec la valeur `1`.

```python
genres = ["roman", "essai", "roman", "poésie", "roman", "essai"]
frequences = {}

# Complétez la boucle.
for genre in genres:
    pass

print(frequences)
```

```python
# Correction
frequences = {}

for genre in genres:
    if genre in frequences:
        frequences[genre] = frequences[genre] + 1
    else:
        frequences[genre] = 1

print(frequences)
```

## Fonction : paramètres et `return`

Écrire `compter_valeur(elements, valeur)` sans utiliser `.count()`. La fonction doit **retourner** le résultat, pas seulement l'afficher.

```python
# Complétez cette fonction.
def compter_valeur(elements, valeur):
    pass
```

```python
# Correction
def compter_valeur(elements, valeur):
    compteur = 0
    for element in elements:
        if element == valeur:
            compteur = compteur + 1
    return compteur

compter_valeur(["rouge", "bleu", "rouge"], "rouge")
```
