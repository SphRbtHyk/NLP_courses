4. Définissez `__repr__()` afin qu'un rectangle de dimensions 8 et 5 s'affiche sous la forme `Rectangle(largeur=8, hauteur=5)`.

```python
class Rectangle:

    def __init__(self, largeur: float, hauteur: float) -> None:
        self.largeur = largeur
        self.hauteur = hauteur

    def __repr__(self) -> str:
        return f"""
        Voici la représentation de mon objet
        C'est un rectangle, avec les caractéristiques:
        Largeur: {self.largeur}
        Hauteur: {self.hauteur}
        """

    def aire(self) -> float:
        """Calcule l'aire d'un rectangle."""
        return self.largeur * self.hauteur

    def perimetre(self) -> float:
        """Calcule le périmètre d'un rectangle."""
        return 2 * self.largeur + 2 * self.hauteur

    def redimensionner(self, coefficient: float) -> bool:
        if coefficient > 0:
            self.largeur = self.largeur * coefficient
            self.hauteur = self.hauteur * coefficient
            return True
        return False
```

```python
rect = Rectangle(largeur=2, hauteur=2)

print(rect)
```

3. Redéfinissez `calculer_prix()` dans `BilletEtudiant` afin d'appliquer la réduction. Redéfinissez aussi `__repr__()` pour faire apparaître le taux de réduction.

```python
class BilletEtudiant(Billet):
    """
    """
    
    def __init__(self,
                 trajet: str,
                 prix_base: float,
                 taux_reduction: float) -> None:

        ###### Initialise toi comme le parent
        super().__init__(trajet, prix_base)
        ######
        
        self.taux_reduction = taux_reduction

    def calculer_prix(self) -> float:
        return self.prix_base * self.taux_reduction

    def __repr__(self) -> str:
        return f"""
Le prix du billet étudiant pour le trajet {self.trajet} est {self.calculer_prix()} euros. Le prix de base est {self.prix_base} euros.
        """

billet_etudiant = BilletEtudiant("Paris-Marseille", 100, 0.5)

print(billet_etudiant)
```
