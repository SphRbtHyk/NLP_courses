2. Ajoutez une méthode `ajouter_occurrences(nombre=1)`. Elle augmente la fréquence seulement si `nombre` est un entier strictement positif. Elle renvoie `True` si l'ajout a été effectué et `False` sinon.

```python
class EntreeLexicale:
    """Définit une classe permettant de représenter une entrée
    lexicale dans un lexique.
    """
    def __init__(self, lemme: str, categorie: str) -> None:
        """Initialise un objet de classe EntreLexicale.

        Args:
            lemme (str): Lemme du mot.
            categorie (str): Catégorie grammaticale du mot.
        """
        self.lemme = lemme
        self.categorie = categorie
        self.frequence = 0

    def ajouter_occurences(self, nombre: int = 1) -> bool:
        """Augmente la fréquence si nombre est un entier strictement
        positive.

        Args:
            nombre (int, optional): La valeur de l'incrément. Defaults to 1.

        Returns:
            bool: Si le nombre a été incrémenté.
        """
        if nombre > 0 and type(nombre) == int:
            self.frequence += nombre
            return True
        return False

    def __repr__(self) -> str:
        return f"Le lemme est {self.lemme}. Il apparaît {self.frequence} fois dans le corpus."
```

3. Ajoutez une méthode `est_rare(seuil=5)` qui renvoie un booléen, puis un `__repr__()` produisant par exemple `EntreeLexicale(chat, nom, fréquence=12)`.

```python
class EntreeLexicale:
    """Définit une classe permettant de représenter une entrée
    lexicale dans un lexique.
    """
    def __init__(self, lemme: str, categorie: str) -> None:
        """Initialise un objet de classe EntreLexicale.

        Args:
            lemme (str): Lemme du mot.
            categorie (str): Catégorie grammaticale du mot.
        """
        self.lemme = lemme
        self.categorie = categorie
        self.frequence = 0

    def ajouter_occurences(self, nombre: int = 1) -> bool:
        """Augmente la fréquence si nombre est un entier strictement
        positive.

        Args:
            nombre (int, optional): La valeur de l'incrément. Defaults to 1.

        Returns:
            bool: Si le nombre a été incrémenté.
        """
        if nombre > 0 and type(nombre) == int:
            self.frequence += nombre
            return True
        return False

    def est_rare(self, seuil: int = 5) -> bool:
        """Retourne vrai si la fréquence du mot est inférieure à seuil."""
        return self.frequence < seuil

    def __repr__(self) -> str:
        return f"Le lemme est {self.lemme}. Il apparaît {self.frequence} fois dans le corpus."
```

```python
entree_lexicale = EntreeLexicale(lemme="chat", categorie="nom")
entree_lexicale.ajouter_occurences(nombre=2)

assert entree_lexicale.est_rare() == True

entree_lexicale.ajouter_occurences(nombre=10)
assert entree_lexicale.est_rare() == False
```

4. Ajoutez les méthodes `duree_restante()` et `progression()`. La première renvoie la somme des durées des tâches non terminées. La seconde renvoie le pourcentage de tâches terminées, ou 0 si le projet est vide. Testez le projet avec au moins quatre tâches.

```python
class Projet:
    def __init__(self, nom: str) -> None:
        self.nom = nom
        self.taches = []

    def ajouter(self, tache: Tache) -> bool:
        for tache_existante in self.taches:
            if tache.titre == tache_existante.titre:
                print(f"Une tache au nom de {tache.titre} existe déjà dans le projet")
                return False
        self.taches.append(tache)
        return True

    def chercher(self, titre: str) -> Union[None, Tache]:
        for tache_existante in self.taches:
            if titre == tache_existante.titre:
                return tache_existante
        return None

    def duree_restante(self) -> float:
        duree_totale = 0
        for tache_existante in self.taches:
            if tache_existante.terminee == False:
                duree_totale += tache_existante.duree
        return duree_totale

    def progression(self) -> float:
        nombre_termine = 0
        for tache_existante in self.taches:
            if tache_existante.terminee:
                nombre_termine += 1
        return nombre_termine/len(self.taches) * 100

projet = Projet(nom="appartement")
menage = Tache(titre="menage", duree=60)

projet.ajouter(menage)

assert projet.duree_restante() == 60 

assert projet.progression() == 0

menage.terminer()

assert projet.progression() == 100
```
