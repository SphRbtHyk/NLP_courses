# TP 4 — Introduction à la programmation objet

Écrivez chaque classe intégralement dans les cellules prévues. Testez les méthodes au fur et à mesure.

## Exercice 1 — Manipuler des rectangles

1. Définissez une classe `Rectangle`. Son constructeur reçoit une largeur et une hauteur et les stocke dans deux attributs. Créez un rectangle de largeur 8 et de hauteur 5, puis affichez ses attributs.

2. Ajoutez les méthodes `aire()` et `perimetre()`. Elles doivent calculer et renvoyer les valeurs correspondantes. Testez-les sur plusieurs rectangles.

3. Ajoutez une méthode `redimensionner(coefficient)`. Si le coefficient est strictement positif, elle multiplie les deux dimensions par celui-ci et renvoie `True`. Sinon, elle ne modifie pas l'objet et renvoie `False`.

4. Définissez `__repr__()` afin qu'un rectangle de dimensions 8 et 5 s'affiche sous la forme `Rectangle(largeur=8, hauteur=5)`.

5. Créez trois rectangles et placez-les dans une liste. Redimensionnez uniquement le deuxième, puis utilisez une boucle pour afficher chaque rectangle et son aire. Vérifiez que les deux autres objets n'ont pas été modifiés.

## Exercice 2 — Construire un petit lexique

1. Définissez une classe `EntreeLexicale`. Son constructeur reçoit un lemme et une catégorie grammaticale. Chaque nouvelle entrée possède aussi une fréquence initiale égale à 0.

2. Ajoutez une méthode `ajouter_occurrences(nombre=1)`. Elle augmente la fréquence seulement si `nombre` est un entier strictement positif. Elle renvoie `True` si l'ajout a été effectué et `False` sinon.

3. Ajoutez une méthode `est_rare(seuil=5)` qui renvoie un booléen, puis un `__repr__()` produisant par exemple `EntreeLexicale(chat, nom, fréquence=12)`.

4. Créez les entrées suivantes et placez-les dans une liste `lexique` : `chat` (nom, 12 occurrences), `dormir` (verbe, 4), `paisible` (adjectif, 2), `maison` (nom, 9) et `observer` (verbe, 6). Les fréquences doivent être ajoutées avec la méthode prévue.

5. Parcourez `lexique` pour calculer la fréquence totale, construire la liste des lemmes rares et trouver l'entrée la plus fréquente. Construisez également une liste contenant uniquement les objets dont la catégorie est `nom`.

## Exercice 3 — Gérer différents types de billets

1. Définissez une classe `Billet` possédant les attributs `trajet` et `prix_base`. Ajoutez une méthode `calculer_prix()` qui renvoie le prix de base et un `__repr__()` lisible.

2. Définissez une classe `BilletEtudiant` qui hérite de `Billet`. Son constructeur reçoit le trajet, le prix de base et un taux de réduction compris entre 0 et 1. Utilisez `super().__init__()` pour initialiser les attributs hérités.

3. Redéfinissez `calculer_prix()` dans `BilletEtudiant` afin d'appliquer la réduction. Redéfinissez aussi `__repr__()` pour faire apparaître le taux de réduction.

4. Créez deux billets ordinaires et deux billets étudiants, puis rassemblez-les dans une même liste. Utilisez une seule boucle pour afficher chaque billet et son prix final.

5. Calculez la recette totale correspondant à la liste de billets. Modifiez ensuite le prix de base d'un seul billet et recalculez la recette afin de vérifier que les autres objets sont restés inchangés.

## Exercice 4 — Organiser les tâches d'un projet

1. Définissez une classe `Tache` avec les attributs `titre`, `duree` et `terminee` (`False` au départ). La méthode `terminer()` passe l'état à `True`. Ajoutez un `__repr__()` indiquant le titre, la durée et l'état de la tâche.

2. Définissez une classe `Projet` avec un nom et une liste de tâches vide. Sa méthode `ajouter(tache)` ajoute la tâche seulement si aucune tâche du projet ne porte déjà le même titre. Elle renvoie un booléen indiquant si l'ajout a réussi.

3. Ajoutez à `Projet` une méthode `chercher(titre)` qui renvoie l'objet correspondant ou `None`, puis une méthode `terminer_tache(titre)` qui utilise `chercher()` et renvoie `True` ou `False` selon que la tâche a pu être terminée.

4. Ajoutez les méthodes `duree_restante()` et `progression()`. La première renvoie la somme des durées des tâches non terminées. La seconde renvoie le pourcentage de tâches terminées, ou 0 si le projet est vide. Testez le projet avec au moins quatre tâches.

5. Écrivez une série de tests avec `assert` couvrant au minimum les cas suivants : projet vide, ajout réussi, refus d'un titre en double, recherche absente, tâche terminée, durée restante et progression après plusieurs modifications.
