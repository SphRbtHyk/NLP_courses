### Exercice 2 — Tri fusion

Le **tri fusion** (*merge sort*) utilise la récursivité et repose sur le principe « diviser pour régner ».

L'algorithme comporte deux opérations :

1. diviser la liste en deux parties ;
2. trier les deux parties puis les **fusionner**.

Par exemple :

```text
[7, 2, 5, 1, 8, 3]

        ↓ division

[7, 2, 5]     [1, 8, 3]

        ↓ divisions successives

[7] [2] [5]   [1] [8] [3]

        ↓ fusions

[2, 7] [5]    [1, 8] [3]

        ↓

[2, 5, 7]     [1, 3, 8]

        ↓

[1, 2, 3, 5, 7, 8]
```
