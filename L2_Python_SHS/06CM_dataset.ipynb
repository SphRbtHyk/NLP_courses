```python
## Question 2
with open("data.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    somme_nb_naissance = 0
    for ligne in lecteur:
        somme_nb_naissance += int(ligne["nb_naissance"])

print(f"Naissances totales: {somme_nb_naissance}")
```

```python
## Question 3
with open("data.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    max_naissance = 0
    prenom_max_donne = ""
    for ligne in lecteur:
        if int(ligne["nb_naissance"]) > max_naissance:
            max_naissance = int(ligne["nb_naissance"])
            prenom_max_donne = ligne["prenom"]

print(f"Le maximum de naissance est {max_naissance} pour le prénom {prenom_max_donne}")


```

```python
## Question 4
#écrire une fonction naissances(prenom) qui renvoie le total pour un
#prénom donné.

def naissances(prenom: str) -> int:
    """Renvoie le total des naissances pour un prénom donné.

    Args:
        prenom (str): Le prénom pour lequel retourner les naissances.

    Returns:
        int: Le nombre de naissance pour le prénom.
    """
    with open("data.csv", encoding="utf-8") as f:
        lecteur = csv.DictReader(f)
        for ligne in lecteur:
            if ligne["prenom"] == prenom:
                return int(ligne["nb_naissance"])


assert naissances("theo") == 10
```
