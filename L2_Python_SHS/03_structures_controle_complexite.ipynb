## Exercice 2 — Examiner une phrase tokenisée

On dispose d'une phrase déjà découpée en mots. La ponctuation a été retirée et tous les mots sont en minuscules.

```python
tokens = ["les", "jeunes", "enfants", "observent", "attentivement", "les", "oiseaux", "migrateurs"]
```

1. Parcourez la liste avec une boucle `for`. Pour chaque mot, affichez le mot et son nombre de caractères.

2. À l'aide d'un accumulateur, calculez le nombre total de caractères, puis la longueur moyenne des mots. N'utilisez pas `sum()`.

3. Construisez une nouvelle liste contenant uniquement les mots d'au moins huit caractères. Affichez aussi le nombre de mots retenus.

4. Comptez le nombre total de voyelles dans la phrase. Utilisez une boucle sur les mots, une boucle sur les lettres et le test d'appartenance à la chaîne `aeiouy`.

5. Construisez un dictionnaire de fréquences : chaque mot doit être une clé et son nombre d'occurrences doit être la valeur associée. Le programme doit fonctionner sans connaître à l'avance les mots présents dans la liste.

```python
tokens = ["les", "jeunes", "enfants", "observent", "attentivement", "les", "oiseaux", "migrateurs"]

# 1. Parcourt la liste et print de chaque mot avec son nombre de caractères
print("Longueur de chaque mot :")
for mot in tokens:
    nombre_caract = len(mot)
    print(f"Le mot '{mot}' fait {nombre_caract} caractères.")

# 2. Calcul du nombre total de caractères
total_caract = 0 # Accumulateur!
for mot in tokens:
    total_caract += len(mot)

moyenne_caract = total_caract / len(tokens)
print(f"\nTotal de caractères : {total_caract}")
print(f"Moyenne : {moyenne_caract:.2f} caractères par mot")

# 3. Filtre des mots d'au moins 8 caractères
mots_longs = []
for mot in tokens:
    if len(mot) >= 8:
        mots_longs.append(mot)

print(f"\nMots longs (≥8 caractères) : {mots_longs}")
print(f"Nombre de mots retenus : {len(mots_longs)}")

# 4. Compte des voyelles
voyelles = "aeiouy"
nombre_voyelles = 0
for mot in tokens: # Double boucle!
    for lettre in mot:
        if lettre in voyelles:
            nombre_voyelles += 1

print(f"\nNombre total de voyelles : {nombre_voyelles}")

# 5. Construction d'un dictionnaire de fréquences
frequence = {}
for mot in tokens:
    if mot in frequence:
        frequence[mot] = frequence[mot] + 1
    else:
        frequence[mot] = 1

print("\nTableau de fréquence des mots :")
for mot, nb in frequence.items():
    print(f"Mot : '{mot}', Nombre d'occurrences : {nb}")
```

## Exercice 3 — Étudier la suite de Syracuse

À partir d'un entier positif `n`, on répète les opérations suivantes :

- si `n` est pair, on le divise par 2 ;
- sinon, on le remplace par `3 * n + 1` ;
- le calcul s'arrête lorsque `n` vaut 1.

1. Pour `n = 27`, écrivez une boucle `while` qui calcule et affiche successivement toutes les valeurs jusqu'à 1.

2. Modifiez le programme pour compter le nombre d'étapes nécessaires et mémoriser la plus grande valeur atteinte. Affichez ces deux résultats à la fin.

3. Stockez maintenant toute la suite dans une liste. Affichez sa longueur, ses cinq premières valeurs et ses cinq dernières valeurs.

4. Calculez le nombre d'étapes pour chacune des valeurs de départ `9`, `27` et `97`. Stockez les résultats dans un dictionnaire dont la valeur de départ est la clé.

5. Sécurisez le programme en imposant une limite de 1 000 étapes. La boucle doit s'arrêter si elle atteint 1 ou si cette limite est dépassée. Affichez ensuite `terminé` ou `interrompu` selon le cas.

```python
# =====================
# 1. Boucle simple pour n = 27
# =====================
n = 27
while n != 1:
    print(n)
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
print("1")
print("\nSuite affichée jusqu'à 1.")

# =====================
# 2. Compte des étapes + stockage du max
# =====================
n = 27
etapes = 0
max_valeur = n
calculer_n = n

while n != 1:
    etapes += 1
    calculer_n = n
    if calculer_n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    if n > max_valeur:
        max_valeur = n

print(f"Nombre d'étapes : {etapes}")
print(f"Plus grande valeur atteinte : {max_valeur}")

# =====================
# 3. Stockage de la suite
# =====================
n = 27
suite = []
etapes = 0

while etapes < 1000:
    suite.append(n)
    if n == 1:
        break
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
    etapes += 1

print(f"Longueur de la suite : {len(suite)}")
print(f"5 premières valeurs : {suite[:5]}")
print(f"5 dernières valeurs : {suite[-5:]}")

# =====================
# 4. Calcul des étapes pour plusieurs valeurs
# =====================
valeurs_dep = [9, 27, 97]
dictionnaire_etapes = {}

for n0 in valeurs_dep:
    n = n0
    etapes = 0
    
    while n != 1:
        etapes += 1
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        if etapes >= 1000:  # sécurité
            break
    
    dictionnaire_etapes[n0] = etapes

print("\nNombre d'étapes pour chaque valeur de départ :")
for n0, nb_etapes in dictionnaire_etapes.items():
    print(f"n = {n0} : {nb_etapes} étapes")

# =====================
# 5. Limite à 1000 étapes
# =====================
for n0 in [9, 27, 97]:
    n = n0
    etapes = 0
    suite = []
    
    while etapes < 1000:
        suite.append(n)
        if n == 1:
            break
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        etapes += 1
    
    suite.append(n)
    
    if n == 1:
        statut = "terminé"
    else:
        statut = "interrompu"
    
    etapes_totales = len(suite)
    print(f"n = {n0} : {etapes_totales} étapes (statut : {statut})")
```

## Exercice 4 — Comparer deux algorithmes de recherche

On cherche une valeur dans une liste triée d'entiers. Chaque fonction devra renvoyer un tuple composé d'un booléen et du nombre de comparaisons effectuées.

1. Écrivez une fonction `recherche_lineaire(liste, cible)` qui examine les éléments depuis le début et renvoie `(True, nombre_comparaisons)` si la cible est trouvée. Si elle est absente, elle doit renvoyer `(False, nombre_comparaisons)`.

2. Écrivez une fonction `recherche_dichotomique(liste, cible)` qui divise par deux la zone de recherche à chaque itération et renvoie le même type de résultat.

3. Testez les deux fonctions sur `list(range(1_000_000))` avec les cibles `0`, `499_999`, `999_999` et `1_000_001`. Présentez le nombre de comparaisons obtenu pour chaque cas.

4. Dans une cellule Markdown, indiquez pour chaque algorithme sa complexité dans le meilleur et dans le pire des cas. Expliquez brièvement pourquoi la position de la cible affecte davantage la recherche linéaire.


5. Importez `time` et mesurez avec `time.perf_counter()` le temps moyen nécessaire pour rechercher une valeur absente dans des listes de tailles `10_000`, `100_000` et `1_000_000`. Répétez chaque recherche vingt fois et ne mesurez pas le temps de création des listes. Comparez les résultats à l'analyse théorique.

```python
# 1. recherche linéaire
def recherche_lineaire(liste, cible):
    """Parcours de la liste du début à la fin, élément par élément."""
    nombre_comparaisons = 0
    for element in liste:
        nombre_comparaisons += 1
        if element == cible:
            return (True, nombre_comparaisons)
    return (False, nombre_comparaisons)


# 2. recherche dichotomique
def recherche_dichotomique(liste, cible):
    """Découpage de la zone de recherche en deux à chaque itération."""
    debut = 0
    fin = len(liste) - 1
    nombre_comparaisons = 0

    while debut <= fin:
        milieu = (debut + fin) // 2
        nombre_comparaisons += 1
        if liste[milieu] == cible:
            return (True, nombre_comparaisons)
        elif liste[milieu] < cible:
            debut = milieu + 1
        else:
            fin = milieu - 1

    return (False, nombre_comparaisons)


# 3. comparaison des deux algorithmes
liste = list(range(1_000_000))
cibles = [0, 499_999, 999_999, 1_000_001]

for cible in cibles:
    resultat_lineaire = recherche_lineaire(liste, cible)
    resultat_dichotomique = recherche_dichotomique(liste, cible)
    print(f"Cible {cible} :")
    print(f"  Linéaire     -> {resultat_lineaire}")
    print(f"  Dichotomique -> {resultat_dichotomique}")
    print()

```

```python
# 5. Mesure des temps (pire des cas)
import time

tailles = [10_000, 100_000, 1_000_000]
nombre_repetitions = 20
cible_absente = -1  # valeur absente de toutes les listes


def temps_moyen(fonction, liste):
    """Mesure du temps moyen d'une recherche répétée sur une même liste."""
    total = 0.0
    for _ in range(nombre_repetitions):
        debut = time.perf_counter()
        fonction(liste, cible_absente)
        total += time.perf_counter() - debut
    return total / nombre_repetitions


# Création des listes AVANT la mesure (hors du chronométrage)
listes = {taille: list(range(taille)) for taille in tailles}

for taille in tailles:
    liste = listes[taille]
    t_lineaire = temps_moyen(recherche_lineaire, liste)
    t_dicho = temps_moyen(recherche_dichotomique, liste)
    print(f"Taille {taille :>9,} :")
    print(f"  Linéaire     -> {t_lineaire * 1000:.3f} ms en moyenne")
    print(f"  Dichotomique -> {t_dicho * 1000:.6f} ms en moyenne")
    print()
```
