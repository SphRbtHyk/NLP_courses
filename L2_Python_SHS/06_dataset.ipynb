4. Donnez la liste des noms et prénoms des individus habitant à Paris.

6. Donnez la liste des noms et prénoms des individus mineurs.

```python
## Question 0: Charger le fichier
import json

with open("data.json") as f:
    dataset = json.load(f)

## Question 1: Donner le nombre moyen d'adresse par personnes
print("---- Exercice 1")
adresse = 0
for info in dataset["personnes"]:
    # Info: type dictionnaire
    adresse += len(info["adresses"])

print(f"Nombre total d'adresses: {adresse}")
print(f"Nombre total de personnes: {len(dataset["personnes"])}")
print(f"Nombre moyen d'adresses par personne: {adresse/len(dataset["personnes"])}")

print("---- Exercice 2")
## Question 2: Donner la liste des noms et prénoms des individus avec plus de deux adresses
adresse = 0

print("Individus avec plus de 2 adresses")
for info in dataset["personnes"]:
    # Info: type dictionnaire
    if len(info["adresses"]) > 2:
        print(info["identite"]["nom"])
        print(info["identite"]["prenom"])

## Question 3: Donner les codes postaux uniques
print("---- Exercice 3")
code_postal = []
for info in dataset["personnes"]:
    for adresse in info["adresses"]:
        code_postal.append(adresse["code_postal"])
print(f"Codes postaux uniques: {set(code_postal)}")

## Question 4: Donner les individus habitant à Paris
print("---- Exercice 4")
for info in dataset["personnes"]:
    for adresse in info["adresses"]:
        if adresse["ville"].lower() == "paris":
            print(f"Nom: {info["identite"]["nom"]}")
            print(f"Prénom: {info["identite"]["prenom"]}")

print("---- Exercice 5")
age = 0
for info in dataset["personnes"]:
    age += info["identite"]["age"]

print(f"Age moyen: {age/len(dataset["personnes"])}")

print("---- Exercice 6")
print("Liste des personnes mineures")
for info in dataset["personnes"]:
    if info["identite"]["age"] < 18:
        print(info["identite"]["nom"], info["identite"]["prenom"])
```

```python
## Question 3 (slides)

def population(nom_commune: str) -> list[int]:
    """Retourne les populations pour nom de commune donné
    (car il peut y avoir plusieurs communes avec le même nom).

    Args:
        nom_commune (str): Nom de la commune concernée.

    Returns:
        int: La population dans la commune.
    """
    url = "https://geo.api.gouv.fr/communes"
    response = requests.get(url, params={"nom": nom_commune})
    output = response.json()
    populations = []
    for info in output:
        populations.append(info["population"])
    return populations

population("Paris")
```
