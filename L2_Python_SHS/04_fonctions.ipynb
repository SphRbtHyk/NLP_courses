**Exercice 2.** 
1. Écrivez une fonction `prix_ttc(prix_ht, tva)` qui reçoit un prix hors taxes et un taux de TVA, et qui **renvoie** le prix TTC. 
2. Rajoutez la gestion des cas limites (TVA négative...)
3. Écrivez grâce à la fonction `assert` trois tests unitaires.


````python
def prix_ttc(prix_ht: float, tva: float) -> float:
    """Retourne le prix TTC étant donné un prix_ht.

    Args:
        prix_ht (float): Le prix HT.
        tva (float): La valeur de la TVA

    Returns:
        float: Le prix taxe comprise.
    """
    if tva <= 0:
        raise ValueError("La TVA ne peut pas être négative!")
    if prix_ht < 0:
        raise ValueError("Le prix HT ne peut pas être négatif")
    return prix_ht + prix_ht * tva

assert prix_ttc(10, 0.2) == 12
````

````python
prix_ttc(10, -0.1)
````

- si `enlever_ponctuation` vaut `True`, les caractères `.,;:!?'"-` sont supprimés;

````python
def nettoyer_texte(
        texte: str,
        majuscules: bool = True,
        enlever_ponctuation: bool =True,
        enlever_trailing_whitespace: bool=True):
    """Nettoie le texte.

    Args:
        texte (str): Le texte à nettoyer.
        majuscules (bool, optional): Si le texte doit être mis en majuscule. Defaults to True.
        enlever_ponctuation (bool, optional): Si la ponctuation doit être enlevée. Defaults to True.
        enlever_trailing_whitespace (bool, optional): Si on doit enlever les espaces de part et d'autre. Defaults to True.
    """
    cleaned_texte = texte
    if enlever_ponctuation:
        cleaned_texte = ""
        ponctuation = ".,;:!?'-"
        for lettre in texte:
            if lettre not in ponctuation:
                cleaned_texte += lettre
    if majuscules:
        cleaned_texte = cleaned_texte.upper()
    if enlever_trailing_whitespace:
        cleaned_texte = cleaned_texte.strip()
    return cleaned_texte


assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=True, enlever_trailing_whitespace=True) == "CEST LA RÉVOLUTION"
assert nettoyer_texte("C'est la révolution.    ", majuscules=False, enlever_ponctuation=True, enlever_trailing_whitespace=True) == "Cest la révolution"
assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=False, enlever_trailing_whitespace=True) == "C'EST LA RÉVOLUTION."
assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=True, enlever_trailing_whitespace=False) == "CEST LA RÉVOLUTION    "

````

````python
def nettoyer_texte(
        texte: str,
        majuscules: bool = True,
        enlever_ponctuation: bool =True,
        enlever_trailing_whitespace: bool=True):
    """Nettoie le texte.

    Args:
        texte (str): Le texte à nettoyer.
        majuscules (bool, optional): Si le texte doit être mis en majuscule. Defaults to True.
        enlever_ponctuation (bool, optional): Si la ponctuation doit être enlevée. Defaults to True.
        enlever_trailing_whitespace (bool, optional): Si on doit enlever les espaces de part et d'autre. Defaults to True.
    """
    cleaned_texte = ""
    ponctuation = ".,;:!?'-"
    for lettre in texte:
        if majuscules:
            lettre = lettre.upper()
        if enlever_ponctuation:
            if lettre not in ponctuation:
                cleaned_texte += lettre
        else:
            cleaned_texte += lettre
    if enlever_trailing_whitespace:
        return cleaned_texte.strip()
    return cleaned_texte


assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=True, enlever_trailing_whitespace=True) == "CEST LA RÉVOLUTION"
assert nettoyer_texte("C'est la révolution.    ", majuscules=False, enlever_ponctuation=True, enlever_trailing_whitespace=True) == "Cest la révolution"
assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=False, enlever_trailing_whitespace=True) == "C'EST LA RÉVOLUTION."
assert nettoyer_texte("C'est la révolution.    ", majuscules=True, enlever_ponctuation=True, enlever_trailing_whitespace=False) == "CEST LA RÉVOLUTION    "
assert nettoyer_texte("C'est la révolution.    ", majuscules=False, enlever_ponctuation=False, enlever_trailing_whitespace=False) == "C'est la révolution.    "
````

**Exercice 2.** 
1. Écrivez une fonction `saisir_nombre(message)` qui :
- affiche `message` et lit la saisie de l'utilisateur avec `input()` ;
- renvoie la saisie convertie en `float` ;
- **tant que** la conversion échoue (`ValueError`), affiche `"Saisie invalide, réessayez."` et redemande : votre fonction ne doit jamais planter, même si l'utilisateur tape `"un"` ou n'importe quoi d'autre.

2. Testez-la avec `saisir_nombre("Entrez un nombre : ")`.


````python
def saisir_nombre(message):
    user_input = None
    while user_input == None:
        try:
            user_input = float(input(message))
            return user_input
        except ValueError:
            print("Vous avez rentré n'importe quoi!")

saisir_nombre("Entre un nombre : ")
````

**Exercice 3.** 
1. Écrivez une fonction `diviser(a, b)` qui renvoie `a / b`, mais qui renvoie `None` si la division est impossible, en utilisant les mots clés `try` / `except`.
2. Vérifiez que `diviser(10, 4)` renvoie `2.5` et que `diviser(3, 0)` renvoie `None` sans erreur.


````python
def diviser(a: float, b: float) -> float | None:
    try:
        return a / b
    except ZeroDivisionError:
        return None


assert diviser(10, 4) == 2.5
assert diviser(3, 0) == None
````

**Exercice 4.** 
1. Écrivez une fonction `moyenne_robuste(liste)` qui renvoie la moyenne des éléments numérique d'une liste;
2. Rajoutez une gestion d'erreur qui permet: d'ignorer silencieusement les éléments d'un autre type (`except TypeError`), et en renvoyant `None` s'il n'y a aucune note valide. 
3. Écrivez et testez les tests unitaires suivants grâce à `assert`:
```python
print(moyenne_robuste([12.0, "absent", 9.5, None]))   # 10.75
print(moyenne_robuste(["a", "b"]))                    # None
print(moyenne_robuste([14.0, "16", 12.0]))            # 13.0
```



````python
from typing import Any


def moyenne_robuste(liste: list[Any]) -> float | None:
    somme = 0
    longueur = 0
    for element in liste:
        if type(element) in [int, float]:
            somme += element
            longueur += 1
    try:
        return somme / longueur
    except ZeroDivisionError:
        return None


print(moyenne_robuste(["a", "b", "absent"]))
````
