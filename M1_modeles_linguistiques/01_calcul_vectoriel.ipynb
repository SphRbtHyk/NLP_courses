NumPy fournit notamment :

## 1. Calculer avec des listes Python

Une liste peut stocker les coordonnées d'un vecteur. Toutefois, les opérateurs des listes n'ont pas le sens des opérations vectorielles.

````python
u = [1, 2, 3]
v = [4, 5, 6]

print(u + v)
print(2 * u)
````

Ici, `+` concatène les listes et `*` répète leur contenu. Pour calculer avec les coordonnées, il faut parcourir les listes.

Exemple d'addition de deux vecteurs de dimension 2 :

````python
a = [2, -1]
b = [3, 4]
somme = []

for i in range(len(a)):
    somme.append(a[i] + b[i])

print(somme)
````

Exemple de multiplication par un scalaire :

````python
a = [2, -1]
scalaire = 3
resultat = []

for coordonnee in a:
    resultat.append(scalaire * coordonnee)

print(resultat)
````

2. Calculez $2u$ puis $-3v$, toujours à l'aide de boucles.

````python
# Vecteur u
u = [1, 0, 3]

# Vecteur v
v = [-2, 5, 1]

# --- Calcul de 2u ---
print("### Calcul de 2u ###")
liste_2u = []
for i in range(len(u)):
    element = u[i] * 2
    liste_2u.append(element)

print(f"  Résultat : {liste_2u}")

# --- Calcul de -3v ---
print("\n### Calcul de -3v ###")
liste_minus_3v = []
for i in range(len(v)):
    element = v[i] * -3
    liste_minus_3v.append(element)

print(f"  Résultat : {liste_minus_3v}")
````

3. Calculez la combinaison linéaire $2u - 3v$.

````python
print("\n### Calcul de 2u-3v ###")

liste_2u_3v = []

for i in range(len(u)):
    liste_2u_3v.append(2 * u[i] - 3 * v[i])

print(f"  Résultat : {liste_2u_3v}")
````

Le produit scalaire s'obtient en multipliant les coordonnées correspondantes, puis en additionnant les produits.

````python
a = [2, -1, 3]
b = [1, 4, 2]
produit = 0

for i in range(len(a)):
    produit = produit + a[i] * b[i]

print(produit)
````

4. Écrivez une fonction `produit_scalaire(u, v)` qui reprend ce calcul et renvoie un nombre. Testez-la sur les vecteurs de l'exemple, puis sur ceux de la première question.

````python
def produit_scalaire(u, v):
    produit = 0
    for i in range(len(v)):
        produit += u[i] * v[i]
    return produit

produit_scalaire(a, b)
produit_scalaire(u, v)
````

## 2. Calculer avec NumPy

`np.array()` transforme une liste en tableau NumPy. Les tableaux numériques contiennent normalement des valeurs d'un même type.

````python
u_np = np.array([1, 0, 3])
v_np = np.array([-2, 5, 1])

print(u_np)
print(type(u_np))
print(u_np.ndim)
print(u_np.shape)
print(u_np.dtype)
````

Pour deux tableaux de même dimension, les opérations arithmétiques sont appliquées coordonnées par coordonnées.

````python
print(u_np + v_np)
print(2 * u_np)
print(u_np * v_np)
print(np.sum(u_np))
print(np.mean(u_np))
````

Le produit `u_np * v_np` est une multiplication coordonnée par coordonnée. Le produit scalaire est calculé avec `np.dot(u_np, v_np)` ou `u_np @ v_np`.

Documentation : [`np.array`](https://numpy.org/doc/stable/reference/generated/numpy.array.html), [`np.dot`](https://numpy.org/doc/stable/reference/generated/numpy.dot.html).

````python
print(np.dot(u_np, v_np))
print(u_np @ v_np)
````

## 3. Implémenter la norme euclidienne

Pour $u = [u_1, …, u_n]$ :

$$\lVert u \rVert = \sqrt{u_1^2 + … + u_n^2} = \sqrt{u \cdot u}.$$

La fonction `sqrt()` du module standard `math` calcule une racine carrée.

````python
from math import sqrt

print(sqrt(25))
````

## 4. Implémenter la similarité cosinus

Pour deux vecteurs non nuls :

$$\cos(u,v) = \frac{u \cdot v}{\lVert u \rVert\,\lVert v \rVert}.$$

La fonction devra renvoyer `None` si l'un des deux vecteurs est nul, car le dénominateur vaudrait alors zéro.

### Exercice 4

1. Écrivez une fonction `cosinus(u, v)` pour des listes Python. Réutilisez `produit_scalaire()` et `norme()` au lieu de recopier leurs calculs.

````python
def cosinus(u, v):
    try:
        return produit_scalaire(u, v)/(norme(u)*norme(v))
    except ZeroDivisionError:
        print("La similarité cosinus est non-définie.")
````

2. Testez la fonction sur les paires suivantes et interprétez les résultats :

- `[1, 0]` et `[4, 0]` ;
- `[1, 0]` et `[0, 3]` ;
- `[1, 2]` et `[-1, -2]` ;
- `[1, 2]` et `[0, 0]`.

````python
print(cosinus([1, 0], [4, 0]))

print(cosinus([1, 0], [0, 3]))

print(cosinus([1, 2], [-1, -2]))

print(cosinus([1, 2], [0, 0]))
````

3. Écrivez `cosinus_numpy(u, v)` pour des tableaux NumPy. Réutilisez `norme_numpy()` et `np.dot()`.

````python
def cosinus_numpy(u, v):
    return np.dot(u, v)/(norme_numpy(u)*norme_numpy(v))
````
