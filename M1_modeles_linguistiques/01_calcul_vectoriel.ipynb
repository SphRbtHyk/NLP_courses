## NumPy en quelques mots

Le principal objet fourni par NumPy est `ndarray`, un tableau de nombres à une ou plusieurs dimensions. Pour ce TP, un tableau à une dimension représentera un vecteur.

NumPy fournit notamment :

- la création de vecteurs et de matrices avec `np.array()` ;
- l'accès aux dimensions et au type des données avec `shape`, `ndim` et `dtype` ;
- les opérations coordonnées par coordonnées, sans écrire explicitement de boucle ;
- l'indexation et les tranches, comme pour les listes Python ;
- des fonctions d'agrégation comme `np.sum()` et `np.mean()` ;
- des fonctions de calcul mathématique et d'algèbre linéaire.

Références : [guide pour débuter](https://numpy.org/doc/stable/user/absolute_beginners.html), [guide de prise en main](https://numpy.org/doc/stable/user/quickstart.html), [documentation de l'API](https://numpy.org/doc/stable/reference/).

## 1. Calculer avec des listes Python

Une liste peut stocker les coordonnées d'un vecteur. Toutefois, les opérateurs des listes n'ont pas le sens des opérations vectorielles.

````python
u = [1, 2, 3]
v = [4, 5, 6]

print(u + v)
print(2 * u)
````

Ici, `+` concatène les listes et `*` répète leur contenu. Pour calculer avec les coordonnées, il faut parcourir les listes.

Exemple d'addition de deux vecteurs de dimension 2 :

````python
a = [2, -1]
b = [3, 4]
somme = []

for i in range(len(a)):
    somme.append(a[i] + b[i])

print(somme)
````

Exemple de multiplication par un scalaire :

````python
a = [2, -1]
scalaire = 3
resultat = []

for coordonnee in a:
    resultat.append(scalaire * coordonnee)

print(resultat)
````

## 2. Calculer avec NumPy

`np.array()` transforme une liste en tableau NumPy. Les tableaux numériques contiennent normalement des valeurs d'un même type.

````python
u_np = np.array([1, 0, 3])
v_np = np.array([-2, 5, 1])

print(u_np)
print(type(u_np))
print(u_np.ndim)
print(u_np.shape)
print(u_np.dtype)
````

Pour deux tableaux de même dimension, les opérations arithmétiques sont appliquées coordonnées par coordonnées.

````python
print(u_np + v_np)
print(2 * u_np)
print(u_np * v_np)
print(np.sum(u_np))
print(np.mean(u_np))
````

Le produit `u_np * v_np` est une multiplication coordonnée par coordonnée. Le produit scalaire est calculé avec `np.dot(u_np, v_np)` ou `u_np @ v_np`.

Documentation : [`np.array`](https://numpy.org/doc/stable/reference/generated/numpy.array.html), [`np.dot`](https://numpy.org/doc/stable/reference/generated/numpy.dot.html).

````python
print(np.dot(u_np, v_np))
print(u_np @ v_np)
````

## 3. Implémenter la norme euclidienne

Pour $u = [u_1, …, u_n]$ :

$$\lVert u \rVert = \sqrt{u_1^2 + … + u_n^2} = \sqrt{u \cdot u}.$$

La fonction `sqrt()` du module standard `math` calcule une racine carrée.

````python
from math import sqrt

print(sqrt(25))
````
